```python
from transformers import pipeline, set_seed

generator = pipeline("text-generation", model="gpt2")
```

```python
prompt = "The old house at the end of the street"

print(prompt)
```

```python
set_seed(10)

output1 = generator(
    prompt,
    max_new_tokens=40,
    temperature=0.3,
    do_sample=True
)

print(output1[0]["generated_text"])
```

```python
set_seed(10)

output2 = generator(
    prompt,
    max_new_tokens=40,
    temperature=1.5,
    do_sample=True
)

print(output2[0]["generated_text"])
```

```python
set_seed(10)

output3 = generator(
    prompt,
    max_new_tokens=40,
    top_k=20,
    do_sample=True
)

print(output3[0]["generated_text"])
```

```python
set_seed(10)

output4 = generator(
    prompt,
    max_new_tokens=40,
    top_p=0.7,
    do_sample=True
)

print(output4[0]["generated_text"])
```

```python
set_seed(10)

output5 = generator(
    prompt,
    max_new_tokens=40,
    repetition_penalty=1.5,
    do_sample=True
)

print(output5[0]["generated_text"])
```

```python
custom_prompt = "One morning, I found a mysterious letter outside my door."

print(custom_prompt)
```

```python
set_seed(1)

custom1 = generator(
    custom_prompt,
    max_new_tokens=50,
    temperature=0.5,
    do_sample=True
)

print(custom1[0]["generated_text"])
```

```python
set_seed(2)

custom2 = generator(
    custom_prompt,
    max_new_tokens=50,
    temperature=1.0,
    do_sample=True
)

print(custom2[0]["generated_text"])
```

```python
set_seed(3)

custom3 = generator(
    custom_prompt,
    max_new_tokens=50,
    temperature=1.5,
    do_sample=True
)

print(custom3[0]["generated_text"])
```

FINAL ANALYSIS
The experiments showed that changing GPT-2's parameters changes the generated text. A low temperature produced more predictable text, while a high temperature produced more creative and random text. Top-k and top-p helped control the possible words selected by the model. Repetition penalty reduced repeated words and phrases. Overall, the parameters control the balance between predictability, creativity, and randomness.
