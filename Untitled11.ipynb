```python
import numpy as np
import matplotlib.pyplot as plt

from tensorflow.keras.datasets import mnist
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

```python
(x_train, y_train), (x_test, y_test) = mnist.load_data()

train = (y_train == 0) | (y_train == 1)
test = (y_test == 0) | (y_test == 1)

x_train = x_train[train]
y_train = y_train[train]

x_test = x_test[test]
y_test = y_test[test]

# Normalize and flatten
x_train = x_train / 255.0
x_test = x_test / 255.0

x_train = x_train.reshape(len(x_train), 784)
x_test = x_test.reshape(len(x_test), 784)
```

```python
model = LogisticRegression(max_iter=1000)

model.fit(x_train, y_train)

prediction = model.predict(x_test)

accuracy = accuracy_score(y_test, prediction)

print("Accuracy:", accuracy)
```

```python
plt.figure(figsize=(10, 3))

for i in range(10):
    plt.subplot(2, 5, i+1)
    plt.imshow(x_test[i].reshape(28,28), cmap="gray")
    plt.title(f"Real: {y_test[i]}\nPred: {prediction[i]}")
    plt.axis("off")

plt.show()
```

```python
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, Dense

input_img = Input(shape=(784,))

encoded = Dense(64, activation="relu")(input_img)
decoded = Dense(784, activation="sigmoid")(encoded)

autoencoder = Model(input_img, decoded)

autoencoder.compile(optimizer="adam", loss="binary_crossentropy")

autoencoder.fit(
    x_train, x_train,
    epochs=5,
    batch_size=256
)
```

```python
generated = autoencoder.predict(x_test[:10])

plt.figure(figsize=(10, 4))

for i in range(10):
    plt.subplot(2, 5, i+1)
    plt.imshow(generated[i].reshape(28,28), cmap="gray")
    plt.axis("off")

plt.show()
```

REFLECTION
The discriminative model learned to identify whether an image is 0 or 1. It learned the differences between the two digits to make predictions. The generative model learned the basic patterns of the digit images and reconstructed similar images. The main difference is that the discriminative model focuses on classification, while the generative model focuses on learning and reproducing the data patterns.
